```python
import random
list1 = [random.randint(0, 100) for i in range(9)]
print("List 1:", list1)

list2 = [random.randint(0, 100) for i in range(9)]
print("List 2:", list2)

additional_list = [a + b for a, b in zip(list1, list2)]
print("Additional List:", additional_list)

extended_list = list1 + list2
print("Extended List 1:", extended_list)

import numpy as np
matrix = np.array(list1).reshape(3,3)

print("\nMatrix :\n", matrix)
submatrix = matrix[0:3:2, [0, 1]]
print("\nSubmatrix :\n", submatrix)

import pandas as pd
df = pd.DataFrame(matrix, columns=['A', 'B', 'C'])
print("\nDataFrame :")
display(df.head())

```
